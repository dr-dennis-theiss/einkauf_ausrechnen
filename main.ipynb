```python
"""
Berechnet die Kosten eines Einkaufs anhand von Produktname,
Einzelpreis und Menge. Das Programm ermittelt den Gesamtpreis
und gibt eine übersichtliche Zusammenfassung aus.
"""

"""
Shopping calculator that determines the total cost of a purchase
based on product name, price, and quantity, then displays
a summary of the result.
"""
```

```python
produktname = "Joghurt"
stueckpreis = 0.49
menge = 10

def einkauf_ausrechnen():
    gesamtpreis = menge * stueckpreis
    print(f"Es wurden {menge} Stück {produktname} gekauft. Gesamtpreis: {gesamtpreis:.2f} €.")

einkauf_ausrechnen()
```
